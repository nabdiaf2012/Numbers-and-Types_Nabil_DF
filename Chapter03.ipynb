```python
## Exercise 3.5
Divide `seconds_per_day` by `seconds_per_hour`. Use floating-point (`/`) division.
    
```

```python
## Exercise 3.6
Divide `seconds_per_day` by `seconds_per_hour`, using integer (`//`) division. Did this number agree with the floating-point value from the previous question, aside from the final .0?
```

```python
seconds_per_day // seconds_per_hour
```

```python
**Answer:** Yes. Integer division (`//`) returns `24`, while floating-point division (`/`) returns `24.0`. The values agree — the only difference is the `.0` decimal suffix, because `//` truncates to an integer and `/` always produces a float.
```
